## Unit 3: Optical Flow Position Hold

__Goal:__ Optical Flow uses a downward facing camera and a downward facing distance sensor for velocity estimation. It can be used to determine speed when navigating without GNSS — in buildings, underground, or in any other GNSS-denied environment.

__Lecture:__  [Optical Flow](./Optical_Flow.pdf) ([Source](https://cvg.ethz.ch/assets/material/visual_computing/exercises/Tutorial5_2023.pdf))

## Unit 4: Relative Position Estimation using AR Tags

__Goal:__ Learn how to use binary square fiducial markers for pose estimation. The main benefit of these markers is that a single marker provides enough correspondences (its four corners) to obtain the camera pose.
